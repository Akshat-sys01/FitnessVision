```python
import json
from pathlib import Path
from sklearn.model_selection import train_test_split

DATASET_DIR = Path("../datasets/MPII Dataset/filtered_exercise_dataset/person_training_dataset")
ANNOTATIONS_FILE = f"{DATASET_DIR}/annotations.json"

TEST_SIZE = 0.20
RANDOM_SEED = 42

# Load annotations
with open(ANNOTATIONS_FILE, "r") as f:
    annotations = json.load(f)

# Get person sample IDs
samples = [
    person["person_id"]
    for person in annotations["annotations"]
]

# Split dataset
train_samples, val_samples = train_test_split(
    samples,
    test_size=TEST_SIZE,
    random_state=RANDOM_SEED,
    shuffle=True
)

# Save split information
split_data = {
    "train": train_samples,
    "validation": val_samples
}

output_file = DATASET_DIR / "dataset_split.json"

with open(output_file, "w") as f:
    json.dump(split_data, f, indent=4)

print("Dataset split completed!")
print(f"Total samples: {len(samples)}")
print(f"Training samples: {len(train_samples)}")
print(f"Validation samples: {len(val_samples)}")
print(f"Training percentage: {len(train_samples)/len(samples)*100:.2f}%")
print(f"Validation percentage: {len(val_samples)/len(samples)*100:.2f}%")
print(f"Saved to: {output_file}")
```
